Pipeline de preparação: descoberta de imagens, redimensionamento, normalização e split treino/validação/teste.

**Aviso:** protótipo acadêmico; não substitui avaliação médica.

```python
# Compatível com Google Colab — descomente se necessário:
# %pip install tensorflow pillow scikit-learn pandas matplotlib

import sys
from pathlib import Path

FASE4 = Path('..').resolve()
if str(FASE4) not in sys.path:
    sys.path.insert(0, str(FASE4))

from src.preprocessamento import (
    DEFAULT_DATASET_DIR,
    DEFAULT_PROCESSED_DIR,
    build_class_index,
    class_distribution,
    discover_images,
    save_splits,
    stratified_split,
    load_image,
)
import matplotlib.pyplot as plt
```

```python
paths, labels = discover_images(DEFAULT_DATASET_DIR)
print(f'Total de imagens: {len(paths)}')
print('Distribuição por classe:')
for cls, n in class_distribution(labels).items():
    print(f'  {cls}: {n}')
```

```python
classes, class_to_idx = build_class_index(labels)
splits = stratified_split(paths, labels)
out_file = save_splits(splits, classes, DEFAULT_PROCESSED_DIR)
print(f'Classes: {classes}')
print(f'Splits salvos em: {out_file}')
for name, split in splits.items():
    print(f'  {name}: {len(split["paths"])} imagens')
```

```python
fig, axes = plt.subplots(1, 5, figsize=(14, 3))
shown = set()
for path, label in zip(paths, labels):
    if label in shown:
        continue
    shown.add(label)
    idx = list(shown).index(label)
    axes[idx].imshow(load_image(path))
    axes[idx].set_title(label, fontsize=8)
    axes[idx].axis('off')
plt.suptitle('Exemplos após pré-processamento (224×224, RGB, [0,1])')
plt.tight_layout()
plt.show()
```
